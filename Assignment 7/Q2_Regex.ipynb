# Question 2: Regex extraction and replacement

Using the raw tickets and Python `re`:
- use `re.findall()` to extract: words longer than 5 letters, numbers (keeping decimals and version numbers such as 95.5 and 2.0.1 as single matches), capitalized words, words beginning with a vowel
- use `re.sub()` to replace: email addresses → `<EMAIL>`, URLs → `<URL>`, phone numbers → `<PHONE>`

Your patterns must also work correctly on the provided `TEST_LINES`.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize, RegexpTokenizer
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, LancasterStemmer, RegexpStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
import string
import re
import pandas as pd

roll_number = "1024170030"
name = "Nikunj"
fingerprint = "E5316233"

tickets = {
    "T01": "The file is 95.5% uploaded.",
    "T02": "The LMS says 'Not Submitted'.",
    "T03": "The VPN crashes while opening. Please help.",
    "T04": "The laptop is showing an error. Please resolve it.",
    "T05": "Password is running slowly after resetting my password. I don't know what changed.",
    "T06": "The Wi-Fi works, but the VPN doesn't.",
    "T07": "Laptop stopped responding after restarting the system. Can you check this?",
    "T08": "Email is not working on my laptop. Please resolve it.",
    "T09": "Printer crashes while opening while uploading the file. Can you check this?",
    "T10": "Python is running slowly before the assignment deadline. I don't know what changed.",
    "T11": "See https://lms.thapar.edu/help for details.",
    "T12": "I cannot access the laptop on my laptop. Please resolve it."
}

stop_words = set(stopwords.words('english'))

def process_ticket(text):
    sentences = sent_tokenize(text)
    tokens = word_tokenize(text)
    punct_tokens = [t for t in tokens if t in string.punctuation]
    stop_tokens = [t for t in tokens if t.lower() in stop_words]
    final_tokens = [t.lower() for t in tokens if t not in string.punctuation and t.lower() not in stop_words]
    return {
        "sentences": sentences,
        "tokens": tokens,
        "punct_tokens": punct_tokens,
        "stop_tokens": stop_tokens,
        "final_tokens": final_tokens
    }
```

```python
TEST_LINES = [
    "Contact me at nikunj.123@thapar.edu for help.",
    "Visit https://lms.thapar.edu/help or http://example.com/path?x=1 for details.",
    "Call me on 9876543210 or +91-9876543210 or (987) 654-3210.",
    "The version is 2.0.1 and the score was 95.5 percent.",
    "Wi-Fi and state-of-the-art equipment isn't working, Email failed Today."
]
```

## Regex patterns

```python
long_word_pattern = r'\b[A-Za-z]{6,}\b'
number_pattern = r'\b\d+(?:\.\d+)+\b|\b\d+\b'
capitalized_pattern = r'\b[A-Z][a-z]*\b'
vowel_word_pattern = r'\b[AEIOUaeiou][A-Za-z]*\b'

email_pattern = r'[\w\.-]+@[\w\.-]+\.\w+'
url_pattern = r'https?://\S+'
phone_pattern = r'(?:\+?\d{1,3}[-\s]?)?(?:\(\d{3}\)[-\s]?|\d{3}[-\s]?)\d{3}[-\s]?\d{4}
```

## Extraction on all tickets

```python
for ticket_id, text in tickets.items():
    print(ticket_id)
    print("  Long words (>5 letters):", re.findall(long_word_pattern, text))
    print("  Numbers:", re.findall(number_pattern, text))
    print("  Capitalized words:", re.findall(capitalized_pattern, text))
    print("  Words starting with vowel:", re.findall(vowel_word_pattern, text))
    print()
```

## Extraction on TEST_LINES

```python
for line in TEST_LINES:
    print(line)
    print("  Long words (>5 letters):", re.findall(long_word_pattern, line))
    print("  Numbers:", re.findall(number_pattern, line))
    print("  Capitalized words:", re.findall(capitalized_pattern, line))
    print("  Words starting with vowel:", re.findall(vowel_word_pattern, line))
    print()
```

## Replacement with re.sub() on tickets

```python
for ticket_id, text in tickets.items():
    replaced = re.sub(email_pattern, "<EMAIL>", text)
    replaced = re.sub(url_pattern, "<URL>", replaced)
    replaced = re.sub(phone_pattern, "<PHONE>", replaced)
    print(ticket_id, "->", replaced)
```

## Replacement with re.sub() on TEST_LINES

```python
for line in TEST_LINES:
    replaced = re.sub(email_pattern, "<EMAIL>", line)
    replaced = re.sub(url_pattern, "<URL>", replaced)
    replaced = re.sub(phone_pattern, "<PHONE>", replaced)
    print(line, "->", replaced)
```
