# Question 5: Corpus statistics

For S1 (all tokens), S2 (tokens without punctuation), S3 (final tokens), S4 (S3 after Porter stemming), calculate total tokens, unique tokens and TTR. Also print the top 10 words of S1 and S3, plot log(rank) vs log(frequency) for S3, and print the top 5 bigrams of S1 and S3.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize, RegexpTokenizer
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, LancasterStemmer, RegexpStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
import string
import re
import pandas as pd

roll_number = "1024170030"
name = "Nikunj"
fingerprint = "E5316233"

tickets = {
    "T01": "The file is 95.5% uploaded.",
    "T02": "The LMS says 'Not Submitted'.",
    "T03": "The VPN crashes while opening. Please help.",
    "T04": "The laptop is showing an error. Please resolve it.",
    "T05": "Password is running slowly after resetting my password. I don't know what changed.",
    "T06": "The Wi-Fi works, but the VPN doesn't.",
    "T07": "Laptop stopped responding after restarting the system. Can you check this?",
    "T08": "Email is not working on my laptop. Please resolve it.",
    "T09": "Printer crashes while opening while uploading the file. Can you check this?",
    "T10": "Python is running slowly before the assignment deadline. I don't know what changed.",
    "T11": "See https://lms.thapar.edu/help for details.",
    "T12": "I cannot access the laptop on my laptop. Please resolve it."
}

stop_words = set(stopwords.words('english'))

def process_ticket(text):
    sentences = sent_tokenize(text)
    tokens = word_tokenize(text)
    punct_tokens = [t for t in tokens if t in string.punctuation]
    stop_tokens = [t for t in tokens if t.lower() in stop_words]
    final_tokens = [t.lower() for t in tokens if t not in string.punctuation and t.lower() not in stop_words]
    return {
        "sentences": sentences,
        "tokens": tokens,
        "punct_tokens": punct_tokens,
        "stop_tokens": stop_tokens,
        "final_tokens": final_tokens
    }
```

```python
import math
import matplotlib.pyplot as plt
from collections import Counter
from nltk.util import bigrams

porter = PorterStemmer()

S1, S2, S3 = [], [], []
for text in tickets.values():
    result = process_ticket(text)
    S1.extend(result["tokens"])
    S2.extend([t for t in result["tokens"] if t not in string.punctuation])
    S3.extend(result["final_tokens"])

S4 = [porter.stem(token) for token in S3]
```

## Total tokens, unique tokens, TTR

```python
def corpus_stats(name, tokens):
    total = len(tokens)
    unique = len(set(tokens))
    ttr = unique / total if total else 0
    return {"Set": name, "Total_Tokens": total, "Unique_Tokens": unique, "TTR": round(ttr, 4)}

stats_df = pd.DataFrame([
    corpus_stats("S1", S1),
    corpus_stats("S2", S2),
    corpus_stats("S3", S3),
    corpus_stats("S4", S4)
])
stats_df
```

## Top 10 words of S1 and S3

```python
s1_top10 = Counter(S1).most_common(10)
s3_top10 = Counter(S3).most_common(10)

print("Top 10 words in S1:")
for word, freq in s1_top10:
    print(f"  {word}: {freq}")

print()
print("Top 10 words in S3:")
for word, freq in s3_top10:
    print(f"  {word}: {freq}")
```

## log(rank) vs log(frequency) for S3

```python
s3_freq = Counter(S3).most_common()
ranks = range(1, len(s3_freq) + 1)
frequencies = [freq for _, freq in s3_freq]

log_ranks = [math.log(rank) for rank in ranks]
log_frequencies = [math.log(freq) for freq in frequencies]

plt.figure(figsize=(7, 5))
plt.plot(log_ranks, log_frequencies, marker='o')
plt.xlabel("log(rank)")
plt.ylabel("log(frequency)")
plt.title("Zipf plot for S3")
plt.show()
```

## Top 5 bigrams of S1 and S3

```python
s1_bigrams = Counter(list(bigrams(S1))).most_common(5)
s3_bigrams = Counter(list(bigrams(S3))).most_common(5)

print("Top 5 bigrams in S1:")
for bigram, freq in s1_bigrams:
    print(f"  {bigram}: {freq}")

print()
print("Top 5 bigrams in S3:")
for bigram, freq in s3_bigrams:
    print(f"  {bigram}: {freq}")
```
