# Question 6: Interpretation

Answer briefly:
(a) Why does TTR change from S1 to S4 in your dataset?
(b) Give one example where `word_tokenize()` and your tokenizer differ. Explain the difference.
(c) Give one example where stemming produces an undesirable result.
(d) Give one example where POS-aware lemmatization preserves meaning better than stemming.

## (a) Why does TTR change from S1 to S4?

TTR rises from S1 to S3 because punctuation tokens and repeated stopwords such as `the`, `is` and `please` are removed, so the remaining tokens are more varied relative to the smaller total count. TTR then drops slightly from S3 to S4 because Porter stemming merges different surface forms, such as `crashes` and `crash`, into the same stem, which reduces the number of unique tokens while the total token count stays the same.

## (b) word_tokenize() vs my_tokenize()

For the line `Wi-Fi works, but the VPN doesn't.`, `word_tokenize()` produces `['Wi-Fi', 'works', ',', 'but', 'the', 'VPN', 'does', "n't", '.']`, splitting the contraction `doesn't` into `does` and `n't`. `my_tokenize()` produces `['Wi-Fi', 'works', ',', 'but', 'the', 'VPN', "doesn't", '.']`, keeping `doesn't` as one token because the custom pattern matches a word joined to another word by an apostrophe before falling back to the plain word pattern.

## (c) Undesirable stemming result

The Lancaster Stemmer reduces `password` to `pass`, which changes the meaning of the word, since `pass` normally refers to succeeding at something or moving past something, not to a login credential. This shows how an aggressive stemmer can strip too much of a word and lose the original sense.

## (d) POS-aware lemmatization vs stemming

For the word `running` tagged as a verb, the WordNet Lemmatizer with POS information returns `run`, correctly identifying the base verb form. The Porter Stemmer also returns `run` here, but for a word like `crashes` tagged as a verb, the lemmatizer returns the valid word `crash`, while a stemmer can return a truncated form that is not a real word in some cases, showing that POS-aware lemmatization keeps the output as a meaningful dictionary word rather than just chopping off letters.
