# Question 3: Custom tokenization

Implement `my_tokenize(text)` using one `RegexpTokenizer` pattern. The pattern must preserve: contractions (isn't), hyphenated words (Wi-Fi, state-of-the-art), decimals (95.5), versions (2.0.1), email addresses, URLs and phone numbers. Run it on `TEST_LINES` and the tickets. Compare its output with `word_tokenize()`. Identify at least three differences and explain them.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize, RegexpTokenizer
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, LancasterStemmer, RegexpStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
import string
import re
import pandas as pd

roll_number = "1024170030"
name = "Nikunj"
fingerprint = "E5316233"

tickets = {
    "T01": "The file is 95.5% uploaded.",
    "T02": "The LMS says 'Not Submitted'.",
    "T03": "The VPN crashes while opening. Please help.",
    "T04": "The laptop is showing an error. Please resolve it.",
    "T05": "Password is running slowly after resetting my password. I don't know what changed.",
    "T06": "The Wi-Fi works, but the VPN doesn't.",
    "T07": "Laptop stopped responding after restarting the system. Can you check this?",
    "T08": "Email is not working on my laptop. Please resolve it.",
    "T09": "Printer crashes while opening while uploading the file. Can you check this?",
    "T10": "Python is running slowly before the assignment deadline. I don't know what changed.",
    "T11": "See https://lms.thapar.edu/help for details.",
    "T12": "I cannot access the laptop on my laptop. Please resolve it."
}

stop_words = set(stopwords.words('english'))

def process_ticket(text):
    sentences = sent_tokenize(text)
    tokens = word_tokenize(text)
    punct_tokens = [t for t in tokens if t in string.punctuation]
    stop_tokens = [t for t in tokens if t.lower() in stop_words]
    final_tokens = [t.lower() for t in tokens if t not in string.punctuation and t.lower() not in stop_words]
    return {
        "sentences": sentences,
        "tokens": tokens,
        "punct_tokens": punct_tokens,
        "stop_tokens": stop_tokens,
        "final_tokens": final_tokens
    }
```

```python
TEST_LINES = [
    "Contact me at nikunj.123@thapar.edu for help.",
    "Visit https://lms.thapar.edu/help or http://example.com/path?x=1 for details.",
    "Call me on 9876543210 or +91-9876543210.",
    "The version is 2.0.1 and the score was 95.5 percent.",
    "Wi-Fi and state-of-the-art equipment isn't working, Email failed Today."
]
```

```python
custom_pattern = r"https?://\S+|[\w\.-]+@[\w\.-]+\.\w+|\+?\d{1,3}[-\s]?\d{3}[-\s]?\d{3}[-\s]?\d{4}|\d+(?:\.\d+)+|\w+(?:[-'’]\w+)+|\w+|[^\w\s]"

custom_tokenizer = RegexpTokenizer(custom_pattern)

def my_tokenize(text):
    return custom_tokenizer.tokenize(text)
```

## Run on TEST_LINES

```python
for line in TEST_LINES:
    print(line)
    print("  my_tokenize:", my_tokenize(line))
    print("  word_tokenize:", word_tokenize(line))
    print()
```

## Run on tickets

```python
for ticket_id, text in tickets.items():
    print(ticket_id)
    print("  my_tokenize:", my_tokenize(text))
    print("  word_tokenize:", word_tokenize(text))
    print()
```

## Comparison

Three differences between `my_tokenize()` and `word_tokenize()`:

1. **Contractions**: `word_tokenize()` splits `isn't` into `is` and `n't`, while `my_tokenize()` keeps `isn't` as a single token because the pattern matches a word, an apostrophe and another word together.

2. **Hyphenated words**: `word_tokenize()` splits `Wi-Fi` and `state-of-the-art` into separate tokens around the hyphens, while `my_tokenize()` keeps each as one token since the pattern joins a word, a hyphen and another word repeatedly.

3. **Emails, URLs and phone numbers**: `word_tokenize()` breaks `nikunj.123@thapar.edu` and `https://lms.thapar.edu/help` into several tokens around the dots, slashes and the @ symbol, while `my_tokenize()` keeps each as a single token because those patterns are matched first in the regex, before the generic word pattern.

4. **Decimals and versions**: both tokenizers keep `95.5` as one token, but `word_tokenize()` splits `2.0.1` into `2.0.1` only in some NLTK versions and into `2.0`, `.`, `1` in others, while `my_tokenize()` always keeps it as a single token due to the repeated-decimal pattern.
