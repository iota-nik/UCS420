## Question 1

Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list `L` (10 numbers in list now).

i. Print `L`

ii. Add two numbers of your choice to `L` — one using `append()` and one using `insert()` at a specific position. Print `L` after each and **explain in a comment what changed**.

iii. Remove two elements from `L` — one using `remove()` and one using `pop()`. Print `L` after each.

iv. Sort `L` in ascending order using `sort()`, then descending order using `sort(reverse=True)`. Print both.

v. Use slicing to print the first three elements and the last three elements of `L` in a single line each.

vi. Use list comprehension to create a new list containing only the elements of `L` that are greater than the average of `L`.

```python
roll_str = str(1024170030)
L = [int(digit) * 10 for digit in roll_str]

# i)
print(L)

# ii)
L.append(101)
print(L)
L.insert(10, 102)
print(L)
# append() adds the element to the end of the list, while insert(index, element) adds the element at that specific index

# iii)
L.remove(101)
print(L)
L.pop()
print(L)

# iv)
L.sort()
print(L)
L.sort(reverse=True)
print(L)

# v)
print(L[:3])
print(L[-3:])

# vi)
avg = sum(L) / len(L)
L2 = [x for x in L if x > avg]
print(L2)
```
