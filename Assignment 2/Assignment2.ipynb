Create a tuple named `scores` containing 8 marks: use the first 8 values from your list `L` in Q1.

## Question 3

Set the random seed to your own roll number using `random.seed(your_roll_number)` before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

ii. Count and print all odd numbers in the list.

iii. Count and print all even numbers in the list.

iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

v. Print the number that occurs most frequently in your list, and how many times it occurs.

````python
import random
from collections import Counter

random.seed(1024170030)

# i)
randomList = [random.randint(100, 900) for i in range(100)]

# ii)
oddNumbers = [x for x in randomList if x % 2 != 0]
print(oddNumbers)
print(len(oddNumbers))

# iii)
evenNumbers = [x for x in randomList if x % 2 == 0]
print(evenNumbers)
print(len(evenNumbers))

# iv)
def isPrime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primeNumbers = [x for x in randomList if isPrime(x)]
print(primeNumbers)
print(len(primeNumbers))

# v)
mostFreq, count = Counter(randomList).most_common(1)[0]
print(mostFreq)
print(count)
````

## Question 4

Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set `A = {digit * 7 for each digit}` and set `B = {digit * 9 for each digit}`. Print both sets.

vi. Find the union of A and B — all unique values across both sets.

vii. Find the intersection of A and B — values common to both.

viii. Find `A - B` and `B - A` separately (not combined) using `difference()`, and explain in one line how this differs from `symmetric_difference()`.

ix. Find the symmetric difference of A and B.

x. Check whether A is a subset of B, and whether B is a superset of A, using `issubset()` and `issuperset()`.

xi. Ask the user to input a value X. Remove X from set A if present using `discard()` (not `remove()`) — explain in one line why `discard()` is safer than `remove()` when you're not sure the value exists.

````python
L = [1, 0, 2, 4, 1, 7, 0, 0]
A = {digit * 7 for digit in L}
B = {digit * 9 for digit in L}
print(A)
print(B)

# vi)
print(A.union(B))

# vii)
print(A.intersection(B))

# viii)
print(A.difference(B))
print(B.difference(A))
# difference() finds elements strictly in one set but not the other, while symmetric_difference() finds elements unique to either set

# ix)
print(A.symmetric_difference(B))

# x)
print(A.issubset(B))
print(A.issuperset(B))

# xi)
X = int(input("Enter a value: "))
A.discard(X)
print(A)
# discard() will not crash the program if the element is not present, while remove() will raise a KeyError
````
