## Question 2

Create a tuple named `scores` containing 8 marks: use the first 8 values from your list `L` in Q1.

i. Find the highest score and its index in the tuple, and the lowest score and how many times it appears.

ii. Reverse the tuple and return the result as a list (tuples themselves cannot be reversed in place — explain why in a one-line comment, referring to tuple immutability).

iii. Ask the user to input a score. Print its first occurrence index if present in the tuple, or a not present message if not.

iv. Attempt to change one element of the tuple directly (e.g., `scores[0] = 100`). Run it, capture the error Python raises, and explain in one line why this happens and how it differs from a list.

v. Unpack the tuple into a first score, a second score, and all remaining scores using the `*` operator in a single line of unpacking.

```python
L = [10, 0, 20, 40, 70, 0, 0, 30, 0]
scores = tuple(L[:8])

# i)
print(max(scores))
print(scores.index(max(scores)))
print(min(scores))
print(scores.count(min(scores)))

# ii)
print(list(reversed(scores)))
# tuples cannot be reversed in place because they are immutable

# iii)
a = int(input("Enter a score: "))
if a in scores:
    print("Index is ", scores.index(a))
else:
    print("Score is not in the tuple")

# iv)
try:
    scores[0] = 1000
except Exception as e:
    print(e)
# this raises a TypeError because tuples are immutable, while lists allow item assignment

# v)
firstScore, secondScore, *remainingScore = scores
print(firstScore)
print(secondScore)
print(remainingScore)
```
